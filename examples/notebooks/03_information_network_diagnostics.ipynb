# Information-flow and network diagnostics

Inspect directional transfer entropy and the latest correlation-distance minimum spanning tree.

```python
import numpy as np

from marketmind.information import mutual_information, shannon_entropy, transfer_entropy
from marketmind.networks import network_snapshot
from marketmind.synthetic import synthetic_market

prices = synthetic_market(periods=800, assets=8, seed=11)
returns = np.log(prices).diff().dropna()
window = returns.tail(252)
```

```python
diagnostics = {
    'SPX entropy (bits)': shannon_entropy(window['SPX'], bins=20),
    'SPX-NDX MI (nats)': mutual_information(window['SPX'], window['NDX'], k=3),
    'SPX to NDX TE': transfer_entropy(window['SPX'], window['NDX'], k=3),
    'NDX to SPX TE': transfer_entropy(window['NDX'], window['SPX'], k=3),
}
diagnostics
```

```python
snapshot = network_snapshot(window, threshold=0.30)
{
    'mean correlation': snapshot.mean_correlation,
    'weighted clustering': snapshot.clustering,
    'MST coherence': snapshot.mst_coherence,
    'MST edges': snapshot.mst.edges(data=True),
}
```
