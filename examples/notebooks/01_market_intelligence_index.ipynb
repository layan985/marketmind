# Market Intelligence Index quickstart

This notebook runs the complete paper-aligned MII pipeline on deterministic synthetic data. No network access is required.

```python
from marketmind import MarketMind, MarketMindConfig
from marketmind.synthetic import synthetic_market

prices = synthetic_market(periods=800, assets=8, seed=42)
prices.tail()
```

```python
config = MarketMindConfig(window=252, step=21, entropy_bins=20, knn_k=3)
result = MarketMind(config).fit_transform(
    prices[['SPX', 'NDX', 'SX5E', 'ES']], network_data=prices
)
result.to_frame().tail()
```

```python
axes = result.to_frame()[['mii', 'memory', 'information', 'connectivity']].plot(
    figsize=(12, 6), title='MarketMind components'
)
axes.set_ylabel('normalized score')
```

Inspect `result.raw_metrics` before interpreting the normalized composite. The package retains raw, normalized, component, and regime layers separately so every transformation is auditable.
