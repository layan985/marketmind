# Cost-aware walk-forward indicator evaluation

Evaluate the paper's nine fixed signals inside causal MII regimes, then compare them with naive baselines.

```python
from marketmind import MarketMind
from marketmind.backtest import WalkForwardEvaluator
from marketmind.indicators import all_signals
from marketmind.robustness import naive_baselines, transaction_cost_sweep
from marketmind.synthetic import synthetic_market

prices = synthetic_market(periods=800, assets=8, seed=7)
mii = MarketMind().fit_transform(
    prices[['SPX', 'NDX', 'SX5E', 'ES']], network_data=prices
)
asset_returns = prices['SPX'].pct_change()
signals = all_signals(prices['SPX'])
```

```python
evaluation = WalkForwardEvaluator(cost_bps=5, slippage_bps=1).evaluate(
    asset_returns, signals, regimes=mii.regimes['regime']
)
evaluation.summary[['category', 'sharpe', 'max_drawdown', 'trades']].head(20)
```

```python
baselines = naive_baselines(asset_returns, signals['sma_50_200'], random_state=0)
transaction_cost_sweep(asset_returns, baselines, costs_bps=[0, 5, 10, 25])
```
